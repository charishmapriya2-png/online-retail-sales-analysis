```python
#  Remove the broken row
print("Before:", len(df))
df = df.dropna(subset=["Quantity", "Price", "InvoiceDate"])
print("After:", len(df))
print("Missing Price left:", df["Price"].isnull().sum())

#  Remove duplicates
print("Duplicates before:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates left:", df.duplicated().sum())
print("Rows now:", len(df))

#. Fix the date type and get the real date range
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
print(df["InvoiceDate"].dtype)
print(df["InvoiceDate"].min(), "to", df["InvoiceDate"].max())

# Separate cancellations from sales
df["Invoice"] = df["Invoice"].astype(str)
cancelled = df[df["Invoice"].str.startswith("C")]
sales = df[~df["Invoice"].str.startswith("C")]
print("Cancelled:", len(cancelled), "| Sales:", len(sales), "| Total:", len(df))
```

```python
print(df[["Quantity", "Price"]].describe())

print("Sales with Quantity <= 0:", (sales["Quantity"] <= 0).sum())
print("Sales with Price <= 0:", (sales["Price"] <= 0).sum())
print(sales[sales["Price"] <= 0].head())

print("Missing Description:", df["Description"].isnull().sum())
print(df[df["Description"].isnull()].head())

print("Missing Customer ID in sales:", sales["Customer ID"].isnull().sum())
print("Revenue share with no Customer ID (%):",
      round(100 * (sales.loc[sales["Customer ID"].isnull(), "Quantity"]
                   * sales.loc[sales["Customer ID"].isnull(), "Price"]).sum()
            / (sales["Quantity"] * sales["Price"]).sum(), 1))
```

```python
#  Keep only real sales: quantity > 0 and price > 0
print("Sales before:", len(sales))
sales = sales[(sales["Quantity"] > 0) & (sales["Price"] > 0)].copy()
print("Sales after:", len(sales))
print("Min Quantity:", sales["Quantity"].min(), "| Min Price:", sales["Price"].min())   # both should be > 0

# Check whether missing descriptions are gone
print("Missing Description left:", sales["Description"].isnull().sum())

# Create a Revenue column
sales["Revenue"] = sales["Quantity"] * sales["Price"]

# Revenue with no Customer ID
no_id = sales["Customer ID"].isnull()
print("Revenue share with no Customer ID (%):",
      round(100 * sales.loc[no_id, "Revenue"].sum() / sales["Revenue"].sum(), 1))

# Customer-level table, used only for customer questions
sales_customers = sales.dropna(subset=["Customer ID"]).copy()
print("Rows with Customer ID:", len(sales_customers))
```

```python
# Non-product stock codes (codes with no digits, like POST or M)
print(sales[~sales["StockCode"].astype(str).str.contains(r"\d")]["StockCode"].value_counts())

# Very low prices
print(sales[sales["Price"] < 0.01][["Invoice", "StockCode", "Description", "Quantity", "Price"]].head())
```

```python
non_product = ["DOT", "POST", "M", "ADJUST", "PADS", "BANK CHARGES", "D", "S"]

# Apply to sales (compare in uppercase so "m" and "M" both match)
print("Sales before:", len(sales))
sales = sales[~sales["StockCode"].astype(str).str.upper().isin(non_product)].copy()
print("Sales after:", len(sales))
print("Non-product codes left:",
      sales["StockCode"].astype(str).str.upper().isin(non_product).sum())

# Apply the same filter to cancellations
cancelled = cancelled[~cancelled["StockCode"].astype(str).str.upper().isin(non_product)].copy()

# Rebuild the customer-level table from the cleaned sales
sales_customers = sales.dropna(subset=["Customer ID"]).copy()
print("Rows with Customer ID:", len(sales_customers))
```

```python
#  Revenue by month
sales["Month"] = sales["InvoiceDate"].dt.to_period("M")
monthly = sales.groupby("Month")["Revenue"].sum().round(0)
print(monthly)

#  Top 10 products by revenue
top_products = (sales.groupby("Description")["Revenue"].sum()
                .sort_values(ascending=False).head(10).round(0))
print(top_products)

#  Cancellations
cancelled_value = (cancelled["Quantity"].abs() * cancelled["Price"]).sum()
print("Cancelled value as % of sales:", round(100 * cancelled_value / sales["Revenue"].sum(), 1))
cancelled_invoices = cancelled["Invoice"].nunique()
sales_invoices = sales["Invoice"].nunique()
print("Cancelled invoices as % of all invoices:",
      round(100 * cancelled_invoices / (cancelled_invoices + sales_invoices), 1))
```

```python
print(sales.groupby("StockCode")["Description"].nunique().sort_values(ascending=False).head(5))
```

```python
# Pick the most common description for each StockCode
names = (sales.groupby("StockCode")["Description"]
         .agg(lambda x: x.value_counts().index[0]))

# Apply it so every row of a product uses the same name
sales["Product"] = sales["StockCode"].map(names)
cancelled["Product"] = cancelled["StockCode"].map(names)

# Check: every StockCode should now have exactly 1 product name
print(sales.groupby("StockCode")["Product"].nunique().max())

# Re-run the top 10 products using the fixed names
top_products = (sales.groupby("Product")["Revenue"].sum()
                .sort_values(ascending=False).head(10).round(0))
print(top_products)
```

```python
# Revenue by country
country = sales.groupby("Country")["Revenue"].sum().sort_values(ascending=False)
print(country.head(5).round(0))
print("UK share of revenue (%):", round(100 * country["United Kingdom"] / country.sum(), 1))

# Top 10% of customers
cust = sales_customers.groupby("Customer ID")["Revenue"].sum().sort_values(ascending=False)
top_n = int(len(cust) * 0.10)
print("Customers:", len(cust), "| Top 10% =", top_n, "customers")
print("Their share of customer revenue (%):", round(100 * cust.head(top_n).sum() / cust.sum(), 1))
```

```python
from google.colab import files

# Clean sales table
sales_export = sales.copy()
sales_export["Month"] = sales_export["InvoiceDate"].dt.strftime("%Y-%m")
sales_export["Customer ID"] = sales_export["Customer ID"].astype("Int64")
sales_export = sales_export[["Invoice", "InvoiceDate", "Month", "StockCode", "Product",
                             "Quantity", "Price", "Revenue", "Customer ID", "Country"]]
sales_export.to_csv("sales_clean.csv", index=False)

# Cancellations table
cancelled["Product"] = cancelled["Product"].fillna(cancelled["Description"])
cancelled_export = cancelled[["Invoice", "InvoiceDate", "StockCode", "Product",
                              "Quantity", "Price", "Country"]].copy()
cancelled_export["CancelledValue"] = cancelled_export["Quantity"].abs() * cancelled_export["Price"]
cancelled_export.to_csv("cancelled_clean.csv", index=False)

print("Sales rows:", len(sales_export), "| Cancelled rows:", len(cancelled_export))

files.download("sales_clean.csv")
files.download("cancelled_clean.csv")
```
